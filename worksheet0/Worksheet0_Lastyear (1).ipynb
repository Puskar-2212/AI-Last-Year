
## SECTION 4.1 - TASK 1: UNIT CONVERSION PROGRAM

```python
def convert_units():
    """
    Converts between different units of measurement.
    Supports: length, weight, volume conversions.
    """
    print("\n" + "="*60)
    print("TASK 1: UNIT CONVERSION PROGRAM")
    print("="*60)

    conversion_types = {
        '1': 'length',
        '2': 'weight',
        '3': 'volume'
    }

    length_conversions = {
        '1': ('meters to feet', lambda x: x * 3.28084),
        '2': ('feet to meters', lambda x: x / 3.28084)
    }

    weight_conversions = {
        '1': ('kilograms to pounds', lambda x: x * 2.20462),
        '2': ('pounds to kilograms', lambda x: x / 2.20462)
    }

    volume_conversions = {
        '1': ('liters to gallons', lambda x: x * 0.264172),
        '2': ('gallons to liters', lambda x: x / 0.264172)
    }

    try:
        print("\nSelect conversion type:")
        print("1. Length")
        print("2. Weight")
        print("3. Volume")

        choice = input("Enter choice (1-3): ").strip()

        if choice not in conversion_types:
            raise ValueError("Invalid choice. Please enter 1, 2, or 3.")

        conversion_type = conversion_types[choice]

        if conversion_type == 'length':
            conversions = length_conversions
        elif conversion_type == 'weight':
            conversions = weight_conversions
        else:
            conversions = volume_conversions

        print(f"\nSelect {conversion_type} conversion:")
        for key, (desc, _) in conversions.items():
            print(f"{key}. {desc}")

        sub_choice = input("Enter choice: ").strip()

        if sub_choice not in conversions:
            raise ValueError(f"Invalid choice for {conversion_type}.")

        value = float(input("Enter value to convert: "))
        description, conversion_func = conversions[sub_choice]
        result = conversion_func(value)

        print(f"\n Result: {value} {description} = {result:.4f}")

    except ValueError as e:
        print(f"✗ Error: {e}")
    except Exception as e:
        print(f" Unexpected error: {e}")
```

```python
convert_units()
```

## SECTION 4.1 - TASK 2: MATHEMATICAL OPERATIONS ON LIST

```python
def math_operations_on_list():
    """
    Performs various mathematical operations on a list of numbers.
    Operations: sum, average, maximum, minimum
    """
    print("\n" + "="*60)
    print("TASK 2: MATHEMATICAL OPERATIONS ON LIST")
    print("="*60)

    def calculate_sum(numbers):
        """Calculate sum of numbers."""
        return sum(numbers)

    def calculate_average(numbers):
        """Calculate average of numbers."""
        if len(numbers) == 0:
            raise ValueError("Cannot calculate average of empty list.")
        return sum(numbers) / len(numbers)

    def find_maximum(numbers):
        """Find maximum value in numbers."""
        if len(numbers) == 0:
            raise ValueError("Cannot find maximum of empty list.")
        return max(numbers)

    def find_minimum(numbers):
        """Find minimum value in numbers."""
        if len(numbers) == 0:
            raise ValueError("Cannot find minimum of empty list.")
        return min(numbers)

    operations = {
        '1': ('Sum', calculate_sum),
        '2': ('Average', calculate_average),
        '3': ('Maximum', find_maximum),
        '4': ('Minimum', find_minimum)
    }

    try:
        print("\nSelect operation:")
        print("1. Sum")
        print("2. Average")
        print("3. Maximum")
        print("4. Minimum")

        choice = input("Enter choice (1-4): ").strip()

        if choice not in operations:
            raise ValueError("Invalid choice. Please enter 1-4.")

        input_str = input("Enter numbers separated by spaces: ").strip()

        if not input_str:
            raise ValueError("No numbers provided.")

        numbers = [float(x) for x in input_str.split()]

        op_name, op_func = operations[choice]
        result = op_func(numbers)

        print(f"\n {op_name}: {result}")

    except ValueError as e:
        print(f" Error: {e}")
    except Exception as e:
        print(f" Unexpected error: {e}")
```

```python
math_operations_on_list()
```

# SECTION 4.2 - LIST MANIPULATION FUNCTIONS

```python
def extract_every_other(lst):
    """Extract every other element from a list, starting from the first."""
    return lst[::2]

def get_sublist(lst, start, end):
    """Return a sublist from start to end (inclusive)."""
    return lst[start:end+1]

def reverse_list(lst):
    """Reverse a list using slicing."""
    return lst[::-1]

def remove_first_last(lst):
    """Remove first and last elements from a list."""
    return lst[1:-1]

def get_first_n(lst, n):
    """Extract first n elements from a list."""
    return lst[:n]

def get_last_n(lst, n):
    """Extract last n elements from a list."""
    return lst[-n:] if n > 0 else []

def reverse_skip(lst):
    """Extract every second element starting from second-to-last, moving backward."""
    return lst[-2::-2]

# Test all list manipulation functions
print("\n" + "="*60)
print("SECTION 4.2: EXERCISE ON LIST MANIPULATION")
print("="*60)

test_list = [1, 2, 3, 4, 5, 6]

print(f"\nOriginal list: {test_list}")
print(f"1. Extract every other: {extract_every_other(test_list)}")
print(f"2. Get sublist (start=2, end=4): {get_sublist(test_list, 2, 4)}")
print(f"3. Reverse list: {reverse_list(test_list)}")
print(f"4. Remove first and last: {remove_first_last(test_list)}")
print(f"5. Get first 3 elements: {get_first_n(test_list, 3)}")
print(f"6. Get last 2 elements: {get_last_n(test_list, 2)}")
print(f"7. Reverse skip: {reverse_skip(test_list)}")
```

#SECTION 4.3 - NESTED LIST FUNCTIONS

```python
def flatten(lst):
    """Flatten a nested list into a single dimension."""
    result = []
    for item in lst:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

def access_nested_element(lst, indices):
    """Access a specific element from a nested list given its indices."""
    current = lst
    for idx in indices:
        current = current[idx]
    return current

def sum_nested(lst):
    """Calculate sum of all elements in a nested list."""
    total = 0
    for item in lst:
        if isinstance(item, list):
            total += sum_nested(item)
        else:
            total += item
    return total

def remove_element(lst, elem):
    """Remove all occurrences of a specific element from a nested list."""
    result = []
    for item in lst:
        if isinstance(item, list):
            result.append(remove_element(item, elem))
        elif item != elem:
            result.append(item)
    return result

def find_max(lst):
    """Find the maximum element in a nested list."""
    max_val = float('-inf')
    for item in lst:
        if isinstance(item, list):
            max_val = max(max_val, find_max(item))
        else:
            max_val = max(max_val, item)
    return max_val

def count_occurrences(lst, elem):
    """Count occurrences of an element in a nested list."""
    count = 0
    for item in lst:
        if isinstance(item, list):
            count += count_occurrences(item, elem)
        elif item == elem:
            count += 1
    return count

def deep_flatten(lst):
    """Flatten a deeply nested list regardless of depth."""
    return flatten(lst)

def average_nested(lst):
    """Calculate average of all elements in a nested list."""
    total = sum_nested(lst)
    count = len(flatten(lst))
    return total / count if count > 0 else 0

# Test all nested list functions
print("\n" + "="*60)
print("SECTION 4.3: EXERCISE ON NESTED LISTS")
print("="*60)

nested_list = [[1, 2], [3, 4], [5]]
nested_list2 = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
nested_list3 = [[1, 2], [3, [4, 5]], 6]
nested_list4 = [[1, 2], [2, 3], [2, 4]]
nested_list5 = [[1, 2], [3, 2], [4, 5]]
nested_list6 = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
nested_list7 = [[1, 2], [3, 4], [5, 6]]

print(f"\n1. Flatten {nested_list}: {flatten(nested_list)}")
print(f"2. Access element at [1, 2] in {nested_list2}: {access_nested_element(nested_list2, [1, 2])}")
print(f"3. Sum of {nested_list3}: {sum_nested(nested_list3)}")
print(f"4. Remove 2 from {nested_list5}: {remove_element(nested_list5, 2)}")
print(f"5. Max in {nested_list3}: {find_max(nested_list3)}")
print(f"6. Count 2 in {nested_list4}: {count_occurrences(nested_list4, 2)}")
print(f"7. Deep flatten {nested_list6}: {deep_flatten(nested_list6)}")
print(f"8. Average of {nested_list7}: {average_nested(nested_list7)}")
```

## Problem 4: Matrix Operations

```python
A =np.array([[3,4],[7,8]])
B =np.array([[5,3],[2,1]])
print("matrix A:\n", A)
print("matrix B:\n", B)
```

```python
A_inv = np.linalg.inv(A)
identity_check = np.dot(A, A_inv)

print("\nA inverse:\n", A_inv)
print("\nA * A_inv:\n", identity_check)
```

```python
AB = np.dot(A, B)
BA = np.dot(B, A)

print("\nA * B:\n", AB)
print("\nB * A:\n", BA)

print("\nIs AB equal to BA?")
print(np.allclose(AB, BA))
```

```python
AB_transpose = (AB).T
BT_AT = np.dot(B.T, A.T)

print("\n(AB)^T:\n", AB_transpose)
print("\nB^T * A^T:\n", BT_AT)

print("\nIs (AB)^T equal to B^T A^T?")
print(np.allclose(AB_transpose, BT_AT))
```

# Solve System of Linear Equations
# 2x − 3y + z = −1
# x − y + 2z = −3
# 3x + y − z = 9

# Represent AX = B

```python
A_sys = np.array([[2, -3, 1],
                  [1, -1, 2],
                  [3,  1, -1]])

B_sys = np.array([[-1],
                  [-3],
                  [9]])

print("\nMatrix A (system):\n", A_sys)
print("\nMatrix B (system):\n", B_sys)
```

# Solve using Inverse Method

```python
A_sys_inv = np.linalg.inv(A_sys)

X = np.dot(A_sys_inv, B_sys)

print("\nSolution using Inverse Method:")
print("x =", X[0][0])
print("y =", X[1][0])
print("z =", X[2][0])
```

# Solve directly using np.linalg.inv

```python
X_direct = np.linalg.inv(A_sys).dot(B_sys)

print("\nSolution using np.linalg.inv directly:")
print(X_direct)
```
